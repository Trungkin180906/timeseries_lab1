```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1. Load dữ liệu và chuyển date thành DatetimeIndex (Monthly)
df = pd.read_csv("../data/raw/a10.csv", parse_dates=['date'], index_col='date')
df.index.freq = 'MS'
```

```python
# 2. Vẽ đồ thị chuỗi thời gian
plt.figure(figsize=(12, 5))
plt.plot(df.index, df['value'], color='#1f77b4', linewidth=1.8, label='Doanh số (a10)')
plt.title('Chuỗi thời gian: Doanh số bán lẻ dược phẩm chống đái tháo đường (1991 - 2008)', fontsize=13)
plt.xlabel('Năm')
plt.ylabel('Doanh số (triệu $)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.tight_layout()

# 3. Xuất biểu đồ ra thư mục reports để nhúng vào docs
plt.savefig('../reports/figures/01_time_series_plot.png', dpi=300)
plt.show()
```
